1. Para um sinal x(t) = cos(2πf t) com duração de 5 s e frequência de amostragem fs = 44.1 kHz, faça o que se pede nos itens abaixo:

(a) Gere gráficos de x(t) no domÍnio do tempo para f ∈ {500, 5000, 10000} Hz.

(b) Escreva um código para tocar os sinais gerados.

(c) Comente os resultados obtidos

```python
import numpy as np #importa a biblioteca de computação numérica
import matplotlib.pyplot as plt #biblioteca que permite a visualização de dados no Python, bem como toda matemática e plotagem de graficos
from IPython.display import Audio #biblioteca que permite a execução de audios no meu notebook.

# Define  os parameters
duration = 5  # seconds
fs = 44100  # sampling frequency in Hz
f_values = [500, 5000, 10000]  # Vetor de frequências in Hz

# Time vector
t = np.linspace(0, duration, int(fs * duration), endpoint=False)
```

### (a) Gerar gráficos de x(t) no domínio do tempo para f ∈ {500, 5000, 10000} Hz.

```python
signals = []

for f in f_values:
    # Generate cosine signal
    x_t = np.cos(2 * np.pi * f * t)
    signals.append(x_t)

    # Plot the signal
    plt.figure(figsize=(12, 4))
    plt.plot(t, x_t)
    plt.title(f'Sinal Cosseno (f = {f} Hz) no Domínio do Tempo')
    plt.xlabel('Tempo (s)')
    plt.ylabel('Amplitude')
    plt.grid(True)
    plt.xlim(0, 0.01) # Show only a small portion to see the waveform
    plt.show()
```

### (b) Escrever um código para tocar os sinais gerados.

```python
for i, f in enumerate(f_values):
    print(f'Reproduzindo sinal com f = {f} Hz:')
    display(Audio(signals[i], rate=fs))
```

### (c) Comentar os resultados obtidos

**Comentários:**

*   **Gráficos no Domínio do Tempo:**
    *   Para $f = 500$ Hz, a oscilação é mais lenta e as "ondas" são mais espaçadas. É possível observar ciclos completos em um curto período de tempo.
    *   Para $f = 5000$ Hz, a oscilação é significativamente mais rápida, e as ondas ficam mais "apertadas". Já é difícil distinguir ciclos individuais no mesmo intervalo de tempo.
    *   Para $f = 10000$ Hz, a oscilação é ainda mais rápida e as ondas são extremamente densas, mostrando a alta frequência do sinal. Visualmente, o sinal parece quase uma faixa sólida devido à alta taxa de mudança dentro de um pequeno intervalo de tempo.

*   **Sinais Sonoros:**
    *   O sinal de 500 Hz será percebido como um tom grave.
    *   O sinal de 5000 Hz será percebido como um tom agudo.
    *   O sinal de 10000 Hz será um tom muito agudo, possivelmente no limite superior da audição humana para algumas pessoas, soando como um "chiado" fino.

Esses resultados demonstram claramente a relação entre a frequência de um sinal e sua percepção visual no domínio do tempo (mais ciclos por unidade de tempo) e auditiva (altura do som - grave para frequências baixas, agudo para frequências altas).
